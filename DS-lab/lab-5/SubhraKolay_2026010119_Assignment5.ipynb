```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

# Read Dataset
df = pd.read_csv("student_performance.csv")
# Q1. Import and Display Dataset
print("Complete DataFrame:")
print(df)
print("\nFirst 5 Rows:")
print(df.head(5))

print("\nLast 3 Rows:")
print(df.tail(3))


```

```python
# Q2. Basic Information of DataFrame
print("\nDataFrame Information:")
df.info()

print("\nShape:", df.shape)
print("Size:", df.size)
print("Columns:", df.columns)
print("Data Types:\n", df.dtypes)
```

```python

# Q3. Create and Modify DataFrame
df["Total_Marks"] = df[
    ["Python_Marks", "AI_Marks", "Network_Marks"]
].sum(axis=1)

df["Average_Marks"] = df[
    ["Python_Marks", "AI_Marks", "Network_Marks"]
].mean(axis=1)

df["Result"] = df["Average_Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

print("\nUpdated DataFrame:")
print(df)




```

```python
# Q5. Statistical Analysis
subjects = ["Python_Marks", "AI_Marks", "Network_Marks"]

print("\nStatistical Analysis:")
for col in subjects:
    print("\n", col)
    print("Sum:", df[col].sum())
    print("Mean:", df[col].mean())
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("Median:", df[col].median())


```

```python
# Q6. Aggregate Function - agg()
print("\nAggregate Statistics:")
print(df[subjects].agg(["min", "max", "mean", "std"]))
```

```python
# Q7. Percentage Change
df = df.sort_values("Student_ID")

df["Percentage_Change"] = df["Python_Marks"].pct_change() * 100

print("\nPercentage Change:")
print(df[["Student_ID", "Python_Marks", "Percentage_Change"]])


```

```python
# Q8. Sorting Using sort_values()
print("\nStudents Sorted by Python Marks Ascending:")
print(df.sort_values("Python_Marks"))

print("\nStudents Sorted by Average Marks Descending:")
print(df.sort_values("Average_Marks", ascending=False))

print("\nSorted by Department and Python Marks:")
print(df.sort_values(
    ["Department", "Python_Marks"],
    ascending=[True, False]
))

print("\nTop 3 Students:")
print(df.nlargest(3, "Average_Marks"))


```

```python
# Q9. Sorting Using sort_index()
df_index = df.set_index("Student_ID")

print("\nSorted by Index Descending:")
print(df_index.sort_index(ascending=False))

print("\nReset Index:")
print(df_index.sort_index(ascending=False).reset_index())


```

```python
# Q10. Data Extraction Using Relational Operators
print("\nPython Marks Greater Than 80:")
print(df[df["Python_Marks"] > 80])

print("\nAttendance Greater Than or Equal to 90:")
print(df[df["Attendance"] >= 90])

print("\nAI Marks Less Than 70:")
print(df[df["AI_Marks"] < 70])

print("\nNetwork Marks Not Equal to 75:")
print(df[df["Network_Marks"] != 75])


```

```python
# Q11. Data Extraction Using Logical Operators
print("\nPython Marks > 75 AND AI Marks > 80:")
print(df[(df["Python_Marks"] > 75) & (df["AI_Marks"] > 80)])

print("\nAttendance > 90 OR Study Hours > 5:")
print(df[(df["Attendance"] > 90) | (df["Study_Hours"] > 5)])

print("\nCSE Students with Average Marks > 75:")
print(df[(df["Department"] == "CSE") & (df["Average_Marks"] > 75)])

print("\nStudents Not Belonging to IT:")
print(df[~(df["Department"] == "IT")])


```

```python
# Q12. Data Extraction Using iloc
print("\nFirst 5 Rows:")
print(df.iloc[:5])

print("\nRows 3 to 7 and Columns 2 to 5:")
print(df.iloc[2:7, 1:5])

print("\nLast 4 Rows and First 3 Columns:")
print(df.iloc[-4:, :3])

print("\nSecond Row Python Marks:")
print(df.iloc[1, df.columns.get_loc("Python_Marks")])


```

```python
# Q13. Line Plot
plt.figure()
plt.plot(df["Student_ID"], df["Python_Marks"], marker="o")
plt.title("Student ID vs Python Marks")
plt.xlabel("Student ID")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()


```

```python
# Q14. Bar Chart
mean_marks = df[subjects].mean()

plt.figure()
mean_marks.plot(kind="bar")
plt.title("Average Marks of Subjects")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.show()


```

```python
# Q15. Histogram
plt.figure()
plt.hist(df["Average_Marks"], bins=5, edgecolor="black")
plt.title("Distribution of Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.show()


```

```python
# Q16. Box Plot
plt.figure()
df[subjects].boxplot()
plt.title("Subject Marks Comparison")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()


```

```python
# Q17. Pie Chart
department_count = df["Department"].value_counts()

plt.figure()
plt.pie(
    department_count,
    labels=department_count.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Department-wise Student Distribution")
plt.show()


```

```python
# Q18. Student Performance Analysis System
print("\nSTUDENT PERFORMANCE ANALYSIS SYSTEM")

# Read CSV Dataset
data = pd.read_csv("student_performance.csv")

# Display Basic Information
print("\nBasic Information:")
data.info()

# Calculate Total and Average Marks
data["Total_Marks"] = data[
    ["Python_Marks", "AI_Marks", "Network_Marks"]
].sum(axis=1)

data["Average_Marks"] = data[
    ["Python_Marks", "AI_Marks", "Network_Marks"]
].mean(axis=1)

# Student with Highest Average Marks
top_student = data.loc[data["Average_Marks"].idxmax()]

print("\nStudent with Highest Average Marks:")
print(top_student)

# Department-wise Average Marks
department_avg = data.groupby("Department")[
    ["Python_Marks", "AI_Marks", "Network_Marks", "Average_Marks"]
].mean()

print("\nDepartment-wise Average Marks:")
print(department_avg)

# Attendance >= 85 and Average Marks >= 75
print("\nStudents with Attendance >= 85 and Average Marks >= 75:")
print(data[
    (data["Attendance"] >= 85) &
    (data["Average_Marks"] >= 75)
])

# Sort by Average Marks
sorted_data = data.sort_values(
    "Average_Marks",
    ascending=False
)

print("\nStudents Sorted by Average Marks:")
print(sorted_data)

# Bar Chart: Department-wise Average Marks
plt.figure()
department_avg["Average_Marks"].plot(kind="bar")
plt.title("Department-wise Average Marks")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.show()

# Histogram: Distribution of Average Marks
plt.figure()
plt.hist(data["Average_Marks"], bins=5, edgecolor="black")
plt.title("Distribution of Student Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.show()

# Final DataFrame
print("\nFinal Student Performance DataFrame:")
print(data[
    [
        "Student_IDS",
        "Name",
        "Department",
        "Total_Marks",
        "Average_Marks"
    ]
])
```
