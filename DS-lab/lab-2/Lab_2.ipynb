```python
n = int(input("Enter a non-negative integer: "))

factorial = 1
counter = n

while counter > 1:
    factorial *= counter
    counter -= 1

print("Factorial of", n, "=", factorial)
```

```python
number = int(input("Enter an integer: "))

value = abs(number)
digits = 0

if value == 0:
    digits = 1
else:
    while value > 0:
        value //= 10
        digits += 1

print("Number of digits =", digits)
```

```python
number = int(input("Enter an integer: "))

original = number
reverse = 0

while number != 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number //= 10

print("Reverse of", original, "=", reverse)
```

```python
number = int(input("Enter an integer: "))

is_prime = True

if number <= 1:
    is_prime = False
else:
    divisor = 2

    while divisor < number:
        if number % divisor == 0:
            is_prime = False
            break
        divisor += 1

if is_prime:
    print("Prime")
else:
    print("Not Prime")
```

```python
x = float(input("Enter x-coordinate: "))
y = float(input("Enter y-coordinate: "))

match (x, y):
    case (0, 0):
        print("Origin")
    case (x, 0) if x > 0:
        print("Positive X-axis")
    case (x, 0) if x < 0:
        print("Negative X-axis")
    case (0, y) if y > 0:
        print("Positive Y-axis")
    case (0, y) if y < 0:
        print("Negative Y-axis")
    case (x, y) if x > 0 and y > 0:
        print("First quadrant")
    case (x, y) if x < 0 and y > 0:
        print("Second quadrant")
    case (x, y) if x < 0 and y < 0:
        print("Third quadrant")
    case (x, y) if x > 0 and y < 0:
        print("Fourth quadrant")
```

```python
secret_number = 27
attempts = 0

while attempts < 5:
    guess = int(input("Guess the number: "))
    attempts += 1

    if guess == secret_number:
        print("Correct!")
        break
    elif guess > secret_number:
        print("Too High")
    else:
        print("Too Low")

else:
    print("Attempts exhausted. Better luck next time!")
```
