```python
import numpy as np
import pandas as pd
from itertools import combinations

# ==========================================
# Q1. Correlation-Based Feature Redundancy
# ==========================================

data = {
    'F1': [10, 20, 30, 40, 50],
    'F2': [12, 24, 36, 48, 60],
    'F3': [50, 40, 30, 20, 10],
    'F4': [15, 25, 20, 35, 30]
}

df = pd.DataFrame(data)

corr_matrix = df.corr()

print("Q1. Correlation Matrix:")
print(corr_matrix)

threshold = 0.90

print("\nRedundant Feature Pairs:")
for col1, col2 in combinations(df.columns, 2):
    corr = corr_matrix.loc[col1, col2]

    if corr > threshold:
        print(col1, "-", col2, "Correlation:", corr)


# ==========================================
# Q2. Distance-Based Feature Similarity
# ==========================================

F1 = np.array([2, 4, 6, 8, 10])
F2 = np.array([3, 5, 7, 9, 11])

euclidean = np.linalg.norm(F1 - F2)
manhattan = np.sum(np.abs(F1 - F2))
cosine = np.dot(F1, F2) / (np.linalg.norm(F1) * np.linalg.norm(F2))

print("\nQ2. Distance-Based Feature Similarity")
print("Euclidean Distance:", euclidean)
print("Manhattan Distance:", manhattan)
print("Cosine Similarity:", cosine)


# ==========================================
# Q3. Comparing Different Distance Metrics
# ==========================================

F1 = np.array([1, 3, 5, 7])
F2 = np.array([2, 3, 6, 8])

euclidean = np.linalg.norm(F1 - F2)
manhattan = np.linalg.norm(F1 - F2, ord=1)

print("\nQ3. Comparing Distance Metrics")
print("Euclidean Distance (L2):", euclidean)
print("Manhattan Distance (L1):", manhattan)

if euclidean < manhattan:
    print("Euclidean distance is smaller")
else:
    print("Manhattan distance is smaller")


# ==========================================
# Q4. Redundancy Between Binary Features
# ==========================================

F1 = np.array([1, 1, 0, 1, 0, 0, 1, 0])
F2 = np.array([1, 0, 0, 1, 1, 0, 1, 0])

n11 = np.sum((F1 == 1) & (F2 == 1))
n00 = np.sum((F1 == 0) & (F2 == 0))
n10 = np.sum((F1 == 1) & (F2 == 0))
n01 = np.sum((F1 == 0) & (F2 == 1))

hamming = np.sum(F1 != F2)

jaccard_similarity = n11 / (n11 + n10 + n01)
jaccard_distance = 1 - jaccard_similarity

smc = (n11 + n00) / len(F1)

print("\nQ4. Binary Feature Measures")
print("Hamming Distance:", hamming)
print("n11:", n11)
print("n00:", n00)
print("n10:", n10)
print("n01:", n01)
print("Jaccard Similarity:", jaccard_similarity)
print("Jaccard Distance:", jaccard_distance)
print("Simple Matching Coefficient:", smc)


# ==========================================
# Q5. Comparing Binary Similarity Measures
# ==========================================

F1 = np.array([1, 0, 1, 1, 0, 1, 0, 0])
F2 = np.array([1, 0, 0, 1, 1, 1, 0, 1])

hamming = np.sum(F1 != F2)

n11 = np.sum((F1 == 1) & (F2 == 1))
n00 = np.sum((F1 == 0) & (F2 == 0))
n10 = np.sum((F1 == 1) & (F2 == 0))
n01 = np.sum((F1 == 0) & (F2 == 1))

jaccard = n11 / (n11 + n10 + n01)
jaccard_distance = 1 - jaccard

smc = (n11 + n00) / len(F1)

print("\nQ5. Binary Similarity Measures")
print("Hamming Distance:", hamming)
print("Jaccard Distance:", jaccard_distance)
print("Simple Matching Coefficient:", smc)

print("Hamming Distance measures mismatched positions.")
print("Jaccard Distance measures dissimilarity between present features.")
print("SMC measures the overall matching positions.")


# ==========================================
# Q6. Identifying Redundant Features
# ==========================================

data = {
    'Study_Hours': [2, 4, 6, 8, 10, 12],
    'Practice_Hours': [3, 6, 9, 12, 15, 18],
    'Sleep_Hours': [8, 7, 8, 6, 7, 8],
    'Screen_Time': [2, 3, 4, 5, 6, 7]
}

df = pd.DataFrame(data)

corr_matrix = df.corr()

print("\nQ6. Correlation Matrix:")
print(corr_matrix)

threshold = 0.95

redundant_pairs = []

for col1, col2 in combinations(df.columns, 2):
    corr = corr_matrix.loc[col1, col2]

    if corr > threshold:
        redundant_pairs.append((col1, col2))
        print("Redundant Pair:", col1, "-", col2)

# Remove redundant features
columns_to_remove = set()

for col1, col2 in redundant_pairs:
    if col1 not in columns_to_remove:
        columns_to_remove.add(col2)

new_df = df.drop(columns=list(columns_to_remove))

print("\nDataFrame After Removing Redundant Features:")
print(new_df)


# ==========================================
# Q7. Correlation Analysis of Student Performance
# ==========================================

data = {
    'Assignment': [40, 50, 60, 70, 80, 90],
    'Internal': [42, 48, 62, 68, 82, 88],
    'Attendance': [60, 65, 70, 75, 80, 85],
    'Project': [30, 45, 50, 65, 75, 85]
}

df = pd.DataFrame(data)

corr1 = df['Assignment'].corr(df['Internal'])
corr2 = df['Attendance'].corr(df['Project'])

print("\nQ7. Correlation Analysis")
print("Assignment and Internal:", corr1)
print("Attendance and Project:", corr2)

if corr1 > corr2:
    print("Assignment and Internal have stronger positive correlation.")
else:
    print("Attendance and Project have stronger positive correlation.")


# ==========================================
# Q8. Finding the Most Similar Feature Pair
# ==========================================

data = np.array([
    [2, 4, 6, 8, 10],
    [3, 6, 9, 12, 15],
    [10, 8, 6, 4, 2],
    [4, 5, 7, 8, 9]
])

features = ['F1', 'F2', 'F3', 'F4']

similarity_matrix = np.zeros((4, 4))

for i in range(4):
    for j in range(4):
        similarity_matrix[i, j] = (
            np.dot(data[i], data[j]) /
            (np.linalg.norm(data[i]) * np.linalg.norm(data[j]))
        )

similarity_df = pd.DataFrame(
    similarity_matrix,
    index=features,
    columns=features
)

print("\nQ8. Cosine Similarity Matrix:")
print(similarity_df)

pairs = []

for i, j in combinations(range(4), 2):
    pairs.append((features[i], features[j], similarity_matrix[i, j]))

best_pair = max(pairs, key=lambda x: x[2])

print("\nMost Similar Feature Pair:", best_pair)


# ==========================================
# Q9. Applying a Correlation Threshold
# ==========================================

corr_data = {
    'F1': [1.00, 0.92, 0.35, 0.20],
    'F2': [0.92, 1.00, 0.40, 0.25],
    'F3': [0.35, 0.40, 1.00, 0.88],
    'F4': [0.20, 0.25, 0.88, 1.00]
}

corr_df = pd.DataFrame(
    corr_data,
    index=['F1', 'F2', 'F3', 'F4']
)

threshold = 0.85

print("\nQ9. Correlation Threshold")

for col1, col2 in combinations(corr_df.columns, 2):
    corr = abs(corr_df.loc[col1, col2])

    if corr > threshold:
        print(col1, "-", col2, "Correlation:", corr)


# ==========================================
# Q10. Effect of Distance on Feature Vectors
# ==========================================

F1 = np.array([2, 5, 8, 10])
F2 = np.array([4, 5, 6, 12])
F3 = np.array([2, 5, 8, 10])

distance1 = np.linalg.norm(F1 - F2)
distance2 = np.linalg.norm(F1 - F3)

print("\nQ10. Effect of Distance")
print("Distance between F1 and F2:", distance1)
print("Distance between F1 and F3:", distance2)

if distance2 == 0:
    print("F1 and F3 have identical feature values.")


# ==========================================
# Q11. Binary Feature Comparison
# ==========================================

F1 = np.array([1, 1, 0, 0, 1, 0, 1, 0])
F2 = np.array([1, 0, 0, 1, 1, 0, 0, 0])

n11 = np.sum((F1 == 1) & (F2 == 1))
n00 = np.sum((F1 == 0) & (F2 == 0))

smc = (n11 + n00) / len(F1)
hamming = np.sum(F1 != F2)

print("\nQ11. Binary Feature Comparison")
print("Both Features 1:", n11)
print("Both Features 0:", n00)
print("Simple Matching Coefficient:", smc)
print("Hamming Distance:", hamming)


# ==========================================
# Q12. Jaccard Similarity for Binary Features
# ==========================================

Product_A = np.array([1, 1, 0, 1, 0, 1, 0, 0, 1])
Product_B = np.array([1, 0, 0, 1, 1, 1, 0, 1, 0])

common = np.sum((Product_A == 1) & (Product_B == 1))

n11 = common
n10 = np.sum((Product_A == 1) & (Product_B == 0))
n01 = np.sum((Product_A == 0) & (Product_B == 1))

jaccard_similarity = n11 / (n11 + n10 + n01)
jaccard_distance = 1 - jaccard_similarity

print("\nQ12. Jaccard Similarity")
print("Common Purchases:", common)
print("Jaccard Similarity:", jaccard_similarity)
print("Jaccard Distance:", jaccard_distance)

if jaccard_similarity >= 0.7:
    print("The products are highly similar.")
else:
    print("The products are not highly similar.")


# ==========================================
# Q13. Comparing Multiple Binary Features
# ==========================================

data = {
    'F1': [1, 1, 0, 1, 0, 1],
    'F2': [1, 1, 0, 1, 0, 1],
    'F3': [0, 1, 1, 0, 1, 0],
    'F4': [1, 0, 1, 0, 1, 1]
}

df = pd.DataFrame(data)

distance1 = np.sum(df['F1'] != df['F2'])
distance2 = np.sum(df['F1'] != df['F3'])

print("\nQ13. Multiple Binary Features")
print("Hamming Distance F1 and F2:", distance1)
print("Hamming Distance F1 and F3:", distance2)

if distance1 < distance2:
    print("F1 and F2 are the most similar pair.")
elif distance2 < distance1:
    print("F1 and F3 are the most similar pair.")
else:
    print("Both pairs have equal similarity.")

print("Hamming distance zero means both features are identical.")


# ==========================================
# Q14. Calculating Minkowski Distance
# ==========================================

F1 = np.array([2, 4, 6, 8])
F2 = np.array([5, 6, 8, 10])

for r in [1, 2, 3]:
    distance = np.sum(np.abs(F1 - F2) ** r) ** (1 / r)
    print("\nMinkowski Distance for r =", r, ":", distance)

print("\nr = 1 gives Manhattan Distance.")
print("r = 2 gives Euclidean Distance.")


# ==========================================
# Q15. Feature Similarity Using Cosine Similarity
# ==========================================

F1 = np.array([1, 2, 3, 4])
F2 = np.array([2, 4, 6, 8])
F3 = np.array([4, 3, 2, 1])

cosine1 = np.dot(F1, F2) / (
    np.linalg.norm(F1) * np.linalg.norm(F2)
)

cosine2 = np.dot(F1, F3) / (
    np.linalg.norm(F1) * np.linalg.norm(F3)
)

print("\nQ15. Cosine Similarity")
print("F1 and F2:", cosine1)
print("F1 and F3:", cosine2)

if cosine1 > cosine2:
    print("F1 and F2 have greater cosine similarity.")
else:
    print("F1 and F3 have greater cosine similarity.")


# ==========================================
# Q16. Identifying Redundancy in a Dataset
# ==========================================

data = {
    'Temperature': [20, 25, 30, 35, 40, 45],
    'Temp_Celsius': [20, 25, 30, 35, 40, 45],
    'Humidity': [60, 65, 55, 70, 50, 75],
    'Pressure': [1000, 1005, 1010, 1015, 1020, 1025]
}

df = pd.DataFrame(data)

corr_matrix = df.corr()

print("\nQ16. Correlation Matrix:")
print(corr_matrix)

print("\nPerfectly Correlated Feature Pairs:")

redundant_pairs = []

for col1, col2 in combinations(df.columns, 2):
    corr = corr_matrix.loc[col1, col2]

    if abs(corr) == 1:
        redundant_pairs.append((col1, col2))
        print(col1, "-", col2, "Correlation:", corr)

# Remove redundant features while keeping one representative
columns_to_remove = set()

for col1, col2 in redundant_pairs:
    if col1 not in columns_to_remove:
        columns_to_remove.add(col2)

new_df = df.drop(columns=list(columns_to_remove))

print("\nDataFrame After Removing Redundant Features:")
print(new_df)
```
